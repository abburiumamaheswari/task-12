```python
import numpy as np

data = np.array([45, 67, 89, 56, 78, 90, 34, 85, 72, 95])

print("Dataset:", data)
```

```python
print("Sum:", np.sum(data))

print("Mean:", np.mean(data))

print("Median:", np.median(data))

print("Minimum:", np.min(data))

print("Maximum:", np.max(data))

print("Standard Deviation:", np.std(data))

print("Variance:", np.var(data))
```

```python
print("Statistical Analysis")

print("Total:", np.sum(data))
print("Average:", np.mean(data))
print("Middle Value:", np.median(data))
print("Lowest Value:", np.min(data))
print("Highest Value:", np.max(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
```

```python
marks = np.array([
    [80, 85, 90],
    [70, 75, 80],
    [90, 95, 85]
])

print("Original Array:")
print(marks)

print("Column-wise Sum:", np.sum(marks, axis=0))

print("Row-wise Sum:", np.sum(marks, axis=1))

print("Column-wise Mean:", np.mean(marks, axis=0))

print("Row-wise Mean:", np.mean(marks, axis=1))
```

```python
print("Subject-wise Maximum:", np.max(marks, axis=0))

print("Subject-wise Minimum:", np.min(marks, axis=0))

print("Subject-wise Average:", np.mean(marks, axis=0))
```

```python
import numpy as np

students = np.array([
    [85, 90, 78],
    [76, 80, 88],
    [92, 95, 89],
    [65, 70, 72],
    [88, 85, 91]
])

print("Student Marks:")
print(students)

print("Total Marks:", np.sum(students, axis=1))
print("Average Marks:", np.mean(students, axis=1))
```

```python
import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()

data = iris.data

print("Iris Dataset:")
print(data)

print("Dataset Shape:", data.shape)
print("Feature Names:", iris.feature_names)
```

```python
print("Total Sum:", np.sum(data))

print("Feature-wise Sum:", np.sum(data, axis=0))

print("Feature-wise Mean:", np.mean(data, axis=0))
```
